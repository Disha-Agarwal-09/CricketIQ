```python
import pandas as pd

match_summary = pd.read_csv("../data/processed/match_summary.csv")

match_summary["date"] = pd.to_datetime(match_summary["date"])
```

```python
def team_form(team_name, last_n=5):

    # Matches involving the team
    team_matches = match_summary[
        (match_summary["team1"] == team_name) |
        (match_summary["team2"] == team_name)
    ].copy()

    if team_matches.empty:
        print("Team not found.")
        return

    # Latest matches first
    team_matches = team_matches.sort_values(
        by="date",
        ascending=False
    )

    recent = team_matches.head(last_n)

    wins = 0

    print("=" * 55)
    print("TEAM FORM REPORT")
    print("=" * 55)

    print(f"\nTeam : {team_name}\n")

    print(f"Last {last_n} Matches\n")

    for i, row in enumerate(recent.itertuples(), start=1):

        opponent = (
            row.team2 if row.team1 == team_name
            else row.team1
        )

        if row.match_won_by == team_name:
            result = "✅ Won"
            wins += 1
        else:
            result = "❌ Lost"

        print(f"{i}. {result} vs {opponent}")

    losses = len(recent) - wins

    print("\n--------------------------")

    print(f"Wins : {wins}")
    print(f"Losses : {losses}")

    print(f"\nWin Percentage : {(wins/len(recent))*100:.1f}%")

    avg_first = recent["first_innings_score"].mean()
    avg_second = recent["second_innings_score"].mean()

    print(f"\nAverage First Innings Score : {avg_first:.1f}")
    print(f"Average Second Innings Score : {avg_second:.1f}")

    print("=" * 55)

    return recent
```

```python
team_form("Chennai Super Kings")
```
