```python
import pandas as pd

match_summary = pd.read_csv("../data/processed/match_summary.csv")

match_summary.head()
```

```python
def head_to_head(team_a, team_b):
    """
    Displays head-to-head statistics between two IPL teams.
    """

    # Filter matches between the two teams
    matches = match_summary[
        (
            (match_summary["team1"] == team_a) &
            (match_summary["team2"] == team_b)
        )
        |
        (
            (match_summary["team1"] == team_b) &
            (match_summary["team2"] == team_a)
        )
    ].copy()

    # If no matches found
    if matches.empty:
        print("No matches found between these teams.")
        return

    # Convert date column to datetime
    matches["date"] = pd.to_datetime(matches["date"])

    # Sort by latest match
    matches = matches.sort_values(by="date", ascending=False)

    # Basic statistics
    total_matches = len(matches)

    team_a_wins = (matches["match_won_by"] == team_a).sum()
    team_b_wins = (matches["match_won_by"] == team_b).sum()

    team_a_win_percentage = (team_a_wins / total_matches) * 100
    team_b_win_percentage = (team_b_wins / total_matches) * 100

    latest_winner = matches.iloc[0]["match_won_by"]

    # Display Results
    print("=" * 55)
    print(f"HEAD TO HEAD : {team_a} vs {team_b}")
    print("=" * 55)

    print(f"\nTotal Matches Played : {total_matches}")

    print(f"\n{team_a} Wins : {team_a_wins}")
    print(f"{team_b} Wins : {team_b_wins}")

    print(f"\n{team_a} Win Percentage : {team_a_win_percentage:.2f}%")
    print(f"{team_b} Win Percentage : {team_b_win_percentage:.2f}%")

    print(f"\nLatest Match Winner : {latest_winner}")

    print("\nLast 5 Match Winners")
    print("-" * 25)

    for i, winner in enumerate(matches["match_won_by"].head(5), start=1):
        print(f"Match {i}: {winner}")

    print("=" * 55)

    return matches
```

```python
head_to_head(
    "Chennai Super Kings",
    "Mumbai Indians"
)
```

```python
head_to_head(
    "Royal Challengers Bangalore",
    "Kolkata Knight Riders"
)
```
