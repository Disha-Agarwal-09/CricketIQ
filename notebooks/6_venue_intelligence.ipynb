```python
import pandas as pd

match_summary = pd.read_csv("../data/processed/match_summary.csv")
```

```python
venues = sorted(match_summary["venue"].unique())

for i, venue in enumerate(venues, 1):
    print(i, "-", venue)
```

```python
def venue_intelligence(venue_name):

    venue_matches = match_summary[
        match_summary["venue"] == venue_name
    ].copy()

    if venue_matches.empty:
        print("❌ Venue not found.")
        return

    matches_played = len(venue_matches)

    avg_first = venue_matches["first_innings_score"].mean()

    avg_second = venue_matches["second_innings_score"].mean()

    highest = venue_matches["first_innings_score"].max()

    lowest = venue_matches["first_innings_score"].min()

    bat_first_win = venue_matches["bat_first_won"].mean() * 100

    chasing_win = venue_matches["chasing_team_won"].mean() * 100

    highest_chase = venue_matches["second_innings_score"].max()

    print("=" * 60)
    print("🏟 VENUE INTELLIGENCE REPORT")
    print("=" * 60)

    print(f"Venue : {venue_name}")

    print(f"\nMatches Played : {matches_played}")

    print(f"\nAverage 1st Innings Score : {avg_first:.2f}")

    print(f"Average 2nd Innings Score : {avg_second:.2f}")

    print(f"\nHighest 1st Innings Score : {highest}")

    print(f"Lowest 1st Innings Score : {lowest}")

    print(f"Highest Successful Chase : {highest_chase}")

    print(f"\nBat First Win % : {bat_first_win:.2f}%")

    print(f"Chasing Win % : {chasing_win:.2f}%")

    print("=" * 60)

    return venue_matches
```

```python
venue_intelligence("Wankhede Stadium")
```
