```python
import pandas as pd

match_summary = pd.read_csv("../data/processed/match_summary.csv")
```

```python
match_summary["toss_winner_won"] = (
    match_summary["toss_winner"] == match_summary["match_won_by"]
)
```

```python
def toss_impact(venue=None):

    df = match_summary.copy()

    if venue:
        df = df[df["venue"] == venue]

    if df.empty:
        print("Venue not found.")
        return

    total_matches = len(df)

    toss_win_match_win = df["toss_winner_won"].sum()

    percentage = (toss_win_match_win / total_matches) * 100

    bat_first = df[
        (df["toss_decision"] == "bat") &
        (df["toss_winner_won"])
    ]

    bowl_first = df[
        (df["toss_decision"] == "field") &
        (df["toss_winner_won"])
    ]

    print("=" * 60)
    print("TOSS IMPACT REPORT")
    print("=" * 60)

    if venue:
        print(f"\nVenue : {venue}")

    print(f"\nMatches : {total_matches}")

    print(f"\nToss Winner Won Match : {percentage:.2f}%")

    print(f"\nWins after choosing BAT : {len(bat_first)}")

    print(f"Wins after choosing FIELD : {len(bowl_first)}")

    print("=" * 60)
```

```python
toss_impact()
```

```python
toss_impact("Wankhede Stadium")
```

```python
toss_impact("Eden Gardens")
```
